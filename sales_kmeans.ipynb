# Sales Data Clustering using K-Means

**Dataset:** `sales_data_sample.csv`  
**Goal:** Group sales records into clusters using K-Means and find the best number of clusters with the Elbow method.

## 1. Imports

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

## 2. Load the dataset

```python
import os

def find_file(name):
    """Look for a dataset in the usual places; in Google Colab, ask for an upload."""
    folders = [".", "data", "/content", "/content/drive/MyDrive", "/mnt/data",
               os.path.expanduser("~/Downloads"), os.path.expanduser("~/Desktop")]
    for folder in folders:
        path = os.path.join(folder, name)
        if os.path.exists(path):
            print(f"Found dataset: {path}")
            return path
    try:
        from google.colab import files   # only exists in Google Colab
        print(f"'{name}' not found. Please upload it now:")
        files.upload()
        if os.path.exists(name):
            return name
    except ImportError:
        pass
    raise FileNotFoundError(
        f"Could not find '{name}'. Put it in the same folder as this notebook and re-run."
    )
```

```python
df = pd.read_csv(find_file("sales_data_sample.csv"), encoding="latin1")
print("Shape:", df.shape)
df.head()
```

## 3. Select features and scale

```python
X = df[["QUANTITYORDERED", "PRICEEACH", "SALES", "MSRP"]].dropna()
X = StandardScaler().fit_transform(X)
print("Data used for clustering:", X.shape)
```

## 4. Elbow method

```python
wcss = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X)
    wcss.append(km.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(range(1, 11), wcss, marker="o")
plt.xlabel("Number of clusters")
plt.ylabel("WCSS")
plt.title("Elbow method")
plt.xticks(range(1, 11))
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Final clustering
Look at the elbow plot above and set `k` to the point where the curve bends.  
(The original script asked with `input()`; a variable is more reliable in a notebook.)

```python
k = 3   # <-- change this to the k you see at the elbow

labels = KMeans(n_clusters=k, random_state=42, n_init=10).fit_predict(X)
print("Records in each cluster:")
print(pd.Series(labels).value_counts().sort_index())
```
